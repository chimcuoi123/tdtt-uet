```python
#bai 3
kytu = input("nhap ky tu: ")
ord_value = ord(kytu)
if ord_value > 65:
    thuonghoa = ord_value - 32
    print("ky tu hoa tuong ung la: ", chr(thuonghoa))
else:
    hoathuong = ord_value + 32
    print("ky tu thuong tuong ung la: ", chr(hoathuong))
```

```python
#bai 6
a, b, c = map(float, input("nhap 3 canh cua tam giac: ").split()) 
def triareacalc(a, b, c):   
    if a < 0 or b < 0 or c < 0:
        raise ValueError("canh tam giac phai la so duong")
    if a + b <= c or a + c <= b or b + c <= a:
        raise ValueError("khong phai 3 canh cua tam giac")
    sperimeter = (a + b + c) / 2
    return (sperimeter * (sperimeter - a) * (sperimeter - b) * (sperimeter - c)) ** 0.5
print(round(triareacalc(a, b, c), 1))
```

```python
tenho = input("nhap ho va ten: ")
dienkethangtruoc = float(input("nhap so dien tieu thu thang truoc(kWh): "))
dienkethangnay = float(input("nhap so dien tieu thu thang nay(kWh): "))
totaldienke = dienkethangnay + dienkethangtruoc
def checksotien(totaldienke):
    if totaldienke <= 50:
        totaltien = totaldienke * 1984
    elif totaldienke <= 100:
        totaltien = 50 * 1984 + (totaldienke - 50) * 2050
    elif totaldienke <= 200:
        totaltien = 50 * 1984 + 50 * 2050 + (totaldienke - 100) * 2380
    elif totaldienke <= 300:
        totaltien = 50 * 1984 + 50 * 2050 + 100 * 2380 + (totaldienke - 200) * 2998
    elif totaldienke <= 400:
        totaltien = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + (totaldienke - 300) * 3350
    else:
        totaltien = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + 100 * 3350 + (totaldienke - 400) * 3460
    return totaltien
print(f"ho va ten: {tenho}")
print(f"so tien phai tra la: {checksotien(totaldienke)}")
```
