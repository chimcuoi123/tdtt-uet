### Problem 4
#### Write a function that returns the count of positive integers k such that 1 <= k < n and k > (n - 1) / 2. (Example: n=9 -> 4).

```python
n = int(input("nhap so: "))
def count_k(n):
    lowerlimit = (n-1)//2
    count = len(range(lowerlimit+1, n))
    return count
dem_so = count_k(n)
print(dem_so)
```

### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
a, b, c = map(float, input("nhap 3 canh cua tam giac: ").split()) 
def triareacalc(a, b, c):   
    if a < 0 or b < 0 or c < 0:
        raise ValueError("canh tam giac phai la so duong")
    if a + b <= c or a + c <= b or b + c <= a:
        raise ValueError("khong phai 3 canh cua tam giac")
    sperimeter = (a + b + c) / 2
    return (sperimeter * (sperimeter - a) * (sperimeter - b) * (sperimeter - c)) ** 0.5
print(triareacalc(a, b, c))
```

### Problem 7
#### Write a function that returns True if a, b, c can be the side lengths of a triangle; otherwise return False.

```python
a, b, c = map(float, input("nhap 3 canh cua tam giac: ").split())
def check_canh(a, b, c):
    if a < 0 or b < 0 or c < 0 or a + b <= c or a + c <= b or b + c <= a or a == 0 or b == 0 or c == 0:
        return False
    else:
        return True
print(check_canh(a, b, c))
```

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
x = input ("nhap 1 ky tu: ")
n = int(input("nhap so nguyen duong n: "))
def shiftchar(x, n):
    ord_x = ord(x)-ord('a')
    new_position = (((ord_x +n)%26)+ord('a'))
    return chr(new_position)
print(shiftchar(x, n))
```

### Problem 9
#### Write a function that converts an uppercase letter to lowercase. If the input is not a single character, raise ValueError.

```python
kytu = input("nhap 1 ky tu in hoa: ")
def uptolow(kytu):
    if len(kytu) != 1:
        raise ValueError("Vui long nhap 1 ky tu duy nhat")
    ord_value = ord(kytu)
    lower_ord_value = ord_value + 32
    return chr(lower_ord_value)
print(uptolow(kytu))
```

### Problem 12
#### Write a function that returns a right isosceles triangle of height n using '*'.
#### Line i (1-indexed) has i '*', joined with newlines.
#### Example with n = 4
#### *
#### **
#### ***
#### ****

```python
n = int(input("nhap so nguyen duong n: "))
def triangle(n):
    for i in range(1, n+1):
        print("*"*i)
triangle(n)
```
